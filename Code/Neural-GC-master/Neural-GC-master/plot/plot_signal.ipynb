```python
from mne_bids import BIDSPath, read_raw_bids
import mne
import matplotlib.pyplot as plt
```

```python
# 1. 读取数据
base_path   = r'G:\DataSet\HUP_iEEG'
HUP_ID      = '116'
RUN         = '01'
TASK        = 'ictal' # 注意修改 
# TASK        = 'interictal' # 注意修改
# 定义 BIDSPath 对象
bids_path = BIDSPath(
    subject     ='HUP' + HUP_ID,
    session     ='presurgery', 
    task        =TASK,
    acquisition ='seeg', # 注意修改 ecog/seeg
    run         =RUN,  
    datatype    ='ieeg',
    suffix      ='ieeg',
    extension   ='.edf',
    root        = base_path  
)

raw = read_raw_bids(bids_path)
raw.load_data()
```

```python
# 假设你想查看这些通道
channels_to_plot = ['RA1', 'RA2', 'RA3', 'RA4', 'RH1', 'RH2', 'RH3', 'RH4']  # 根据你的数据修改通道名

# 1. 选择特定通道
raw_selected = raw.copy().pick_channels(channels_to_plot)


# 2. 可视化选定通道
raw_selected.plot(scalings='auto', title='Selected Channels', show=True)
```
